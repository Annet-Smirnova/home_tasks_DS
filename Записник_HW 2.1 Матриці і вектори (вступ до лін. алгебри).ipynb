```python
import numpy as np
import random
```

```python
#1 спосіб - Генерація випадкових чисел з модулем random.uniform.
vector = np.array([random.uniform(0, 1) for i in range(10)])

print(vector)
```

```python
#2 спосіб - Створення вектора з допомогою метода numpy np.random.rand
vector2 = np.random.uniform(0, 1, 10)

print(vector2)
```

```python
# Створення матриці matrix
matrix = np.array([[10, 14, 17],
                   [20, 23, 26],
                   [29, 32, 35]])
```

```python
print("\nВектор-рядок 1сп:")
print(vector)
print("\nРозмірність:")
print(vector.shape)

print("\nВектор-рядок 2сп:")
print(vector2)
print("\nРозмірність:")
print(vector2.shape)

print("\nМатриця matrix:")
print(matrix)
print("\nРозмірність:")
print(matrix.shape)
```

```python
# Операцію numpy.expand_dims над вектором 
column_vector = np.expand_dims (vector, axis=1)

print("\n3.Операцію numpy.expand_dims над вектором :")
print("\nПочатковий вектор:")
print(vector)
print("\nПочаткова розмірність:")
print(vector.shape)
print("\nНовий вектор:")
print(column_vector)
print("\nНова розмірність:")
print(column_vector.shape)
```

 Початковий вектор мав одновимірний, його розмірність (10,), операція np.expand_dims додає один вимір до нашого, тобто в результаті отриауємо двовимірний вектор з розмірністю (10, 1) який вже можна буде транспонувати. 
 Аргумент axis=1 - додає вісь після наявної вісі, тобто ми отримали вектор стовбець, якби використали аргументом axis=0 - ми б додали вісь до наявної і отримали б вектор рядок з розмірністю (1, 10)

```python
#транспонування column_vector
column_vector_T = column_vector.T

print("\n4.Транспонування column_vector:")
print("\nПочатковий вектор:")
print(column_vector)
print("\nПочаткова розмірність:")
print(column_vector.shape)
print("\nНовий транспонований вектор:")
print(column_vector_T)
print("\nНова розмірність:")
print(column_vector_T.shape)
```
