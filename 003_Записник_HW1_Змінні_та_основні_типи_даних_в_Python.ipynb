**2. Арифметичні операції**

Створіть дві змінні: `var1` зі значенням 15, `var2` зі значенням 4.

Обчисліть та виведіть (кожен пункт оцінюється в 0.5 бала):
- Суму
- Різницю
- Добуток
- Ділення
- Цілочисельне ділення
- Залишок від ділення
- Піднесення до степеня


```python
var1=15
var2=4
```

```python
summ=var1+var2
rizn=var1-var2
dob=var1*var2
dil=var1/var2
dil_tsile=var1//var2
dil_zal=var1%var2
stepin=var1**var2
```

```python
#Сумма
print(f"Сумма = {summ}")
#Різниця
print(f"Різниця = {rizn}")
#Добуток
print(f"Добуток = {dob}")
#Ділення
print(f"Ділення = {dil}")
#Цілочисельне ділення
print(f"Цілочисельне ділення = {dil_tsile}")
#Залишок від ділення
print(f"Залишок від ділення = {dil_zal}")
#Піднесення до степеня
print(f"Піднесення до степеня = {stepin}")
```
